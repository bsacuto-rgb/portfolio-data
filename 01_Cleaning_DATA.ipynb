# Nous allons travailler sur la meteo.

### D'abord nous allons verifié nos URL

#### Ici nous allons mettre nos import:

```python
import pandas as pd
```

Url des données energies:

```python
url_energy = "https://raw.githubusercontent.com/owid/energy-data/master/owid-energy-data.csv"
df_energy = pd.read_csv(url_energy)
```

Chargement des données CO2/climats


```python
url_co2 = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df_co2 = pd.read_csv(url_co2)
```

```python
print(f"Taille Energy: {df_energy.shape}")
print(f"Taille CO2: {df_co2.shape}")
```

### Explorons nos données Brut.

```python
list(df_energy.columns)
```

```python
df_energy.head()
```

On choisit les colonnes importantes du data frame Energy et on crée un nouveau dataframe a partir de celui_ci :

```python
df_energy_columns = ["country","year","iso_code","population","gdp","fossil_share_energy","coal_share_energy","renewables_share_energy", "low_carbon_share_energy","energy_per_capita","energy_per_gdp"]
```

```python
df_energy_clean =  df_energy[df_energy_columns].copy()
```

On garde nos données seulement à partir des années 90.

```python
df_energy_clean = df_energy_clean[df_energy_clean["year"]>=1990]
```

```python
df_energy_clean.head()
```

```python
df_energy_clean.isnull().sum()
```

On vire les régions avec un ISO null. Un ISO Null veut dire que c'est une région du monde et cela ne nous interesse pas.

```python
df_energy_clean= df_energy_clean.dropna(subset=["iso_code"])
```

```python
print(df_energy_clean.shape)
print(df_energy_clean.isnull().sum())
```

On a encore trop de Nan pour faire un processus de Machine Learning fiable. on va suprimmé les lignes Nan de fossil_share_energy et "renewable_share_energy

```python
df_energy_clean = df_energy_clean.dropna(subset=["fossil_share_energy","renewables_share_energy"])

# On vérifie la taille restante

print (f"Nouvelle taille : {df_energy_clean.shape}")
```

```python
print(df_energy_clean.isnull().sum())
```

Il reste des Nan. on va aussi les suprimmé comme cela notre Dataframe est totalement propre.

```python
df_energy_clean = df_energy_clean.dropna(subset=["gdp"])
print(f"taille finale : {df_energy_clean.shape}")
```

Nous avons un Dataframe Dense pour df_clean_energy. Nous passons au nettoyage du dataset df_co2.

```python
list(df_co2.columns)
```

On choisit les colonens qui nous interessent le plus :

```python
cols_co2 = [
    'country', 'year', 'iso_code', 'co2', 'co2_per_capita', 
    'coal_co2', 'gas_co2', 'oil_co2', 'share_global_co2', 
    'temperature_change_from_co2'
]

df_co2_clean = df_co2[cols_co2].copy()
```

On va faire notre Merge sur le Iso_code et l'année pour evité les erreurs de frappe dans le nom des pays.

```python
# On nettoie les nan sur la colonne iso_code
df_co2_clean = df_co2_clean.dropna(subset=['iso_code'])

# On nettoie les lignes sans données CO2 car c'est le fond de notre sujet
df_co2_clean = df_co2_clean.dropna(subset=["co2","co2_per_capita"])

# On supprimme aussi les années en dessous de 1990 comme on l'a fait sur df_energy_clean
df_co2_clean = df_co2_clean[df_co2_clean["year"]>=1990]

print(f" Taille de notre dataset CO2 : {df_co2_clean.shape}")
```

```python
df_co2_clean.isnull().sum()
```

# On regarde les pays qui ont Nan pour gas et coal et voire si cela s'explique ou si c'est un manque de donnée.

```python
pays_manquants = df_co2_clean[df_co2_clean["coal_co2"].isna()]["country"].unique()
print(f"Nombre de pays avec données manquantes : {len(pays_manquants)}")
print(pays_manquants)
```

Nous pouvons voire que la majorité des pays listé entre dans deux catégories. Des pays porté principalement sur le petrole pour son energie ou des pays comme le Tchad ou l'Angola le suivis statistique peut être complexe faute de moyen. Je decide donc pour gardé l'impact de ces pays dans mon ML de mettre ces données manquantes à 0.

```python
df_co2_clean[["coal_co2","gas_co2","oil_co2"]] = df_co2_clean[["coal_co2","gas_co2","oil_co2"]].fillna(0)
```

On lance le merge :

```python
df_final = pd.merge(
    df_energy_clean,
    df_co2_clean,
    on=["iso_code","year"],
    how="inner"
)
```

On suprimme une des colonnes country car elles est apparue en double

```python
df_final["country"] = df_final["country_x"]
df_final = df_final.drop(columns=["country_x","country_y"])
```

```python
# 5. Le verdict
print(f"\nShape final du dataset : {df_final.shape}")
df_final.head()
```

```python
cols = list(df_final.columns)
cols.remove("country")
cols.insert(0, "country")
df_final = df_final[cols]

df_final.head()
```

```python
# Sauvegarde du dataset propre pour la suite du projet
df_final.to_csv('energy_climat_clean.csv', index=False)
print("Fichier sauvegardé avec succès !")
```
